## 4. グラフを1枚出す

**日本語が読めるかどうかの確認**を兼ねる。

```python
plt.figure(figsize=(6, 3.5))
plt.hist(df["体重g"].dropna() / 1000, bins=30)
plt.xlabel("体重（kg）")
plt.ylabel("種の数")
plt.title("霊長類の体重の分布：体重の値がある265種")
```

### 確認

- [ ] グラフが表示された
- [ ] **軸のラベルとタイトルの日本語が読める**（□□□ になっていない）

**□□□ になった人は、2 のセルをもう一度実行してから、ここをやり直す。**

なお、このグラフは**右に長く伸びた形**をしている。ほとんどの種は軽く、ごく一部だけがとても重い。

このグラフが描いているのは、**体重の値がある265種**である（配布データ全体は376種。111種は体重が空欄）。

実際に計算すると、**平均5.9kg、中央値3.0kg、最大149.3kg**（ゴリラ）である。

**この分布では、一部の大きな値に引っぱられて、平均が中央値より大きくなる。**
平均も分布の位置を表す立派な指標である。ただし、
**「小さい順に並べたときのまん中」を知りたいなら、見るのは中央値である。**
