## 1. 装飾は、何のために使うのか

**答え：読む人の注意を、見てほしいところへ導くため。**それ以外に使わない。

第9回で見た「2年生だけ赤にする」を思い出す。あれは**注意を導く**操作だった。
同時に、**どこへ導くかを決める**操作でもあった。だから、

> **強調するということは、主張することである。**

だから強調は**1ページに1〜2か所**にする。全部を強調すると、どれも強調にならない。

### 足してよい装飾／足してはいけない装飾

| 足してよい | 足してはいけない |
|---|---|
| いちばん言いたい数字を太字にする | 立体的な影・背景画像・回転 |
| 図に矢印で「ここ」と示す | 意味のない色分け（きれいだから） |
| 限界の段落を囲んで目立たせる | 読みにくいフォント |

## 2. 図を入れて、キャプションを直す

第9回で書き出したグラフを、左のフォルダアイコン（📁）にアップロードしてから、下を実行する。

```python
# 図のファイル名とキャプションを、自分のものに直す
figures = [
    ("graph1.png", "図1　学年別の満足度の平均（n＝40）"),
    # ("graph2.png", "図2　学年別・性別の満足度（n＝40）"),
]

html = open("report.html", encoding="utf-8").read()
figs = "\n".join(f'<figure><img src="{s}" alt="{c}"><figcaption>{c}</figcaption></figure>'
                 for s, c in figures)
html = re.sub(r'<div data-fill="figures">.*?</div>',
              f'<div data-fill="figures">{figs}</div>', html, count=1, flags=re.S)
open("report.html", "w", encoding="utf-8").write(html)
display(HTML(html))
```

**キャプションには必ず n（件数）を書く。**
「図1」だけでは、読む人はその図が何人分の話なのか分からない。

## 3. 公開前チェック

**公開すると、取り消せない。**出す前に確かめる。

下のセルは、よくある問題を機械的に探す。
**ただし、機械が見つけられるのはここまでである。**最後は自分の目で読む。

```python
def check(path="report.html"):
    html = open(path, encoding="utf-8").read()
    html = re.sub(r"<!--.*?-->", "", html, flags=re.S)   # コメントは判定から外す
    text = re.sub(r"<.*?>", "", html)
    ng = []

    if "ここに書く" in html or "ここに読み取り" in html:
        ng.append("書いていない場所が残っている")
    if "◯" in text:
        ng.append("◯（伏せ字）が残っている。班名・日付・件数を入れる")
    if re.search(r"n\s*[＝=]\s*\d", text) is None:
        ng.append("図のキャプションに n（件数）が書かれていない")
    if len(re.sub(r"\s", "", re.search(r"data-fill=\"limits\">(.*?)</div>", html, re.S).group(1))) < 30:
        ng.append("「限界」が短すぎる（または空）")
    if re.search(r"\.csv|回答一覧|生データ", text):
        ng.append("データファイルへの言及がある。生データは載せない")
    if re.search(r"[一-龥]{1,3}(さん|くん|君|様)", text):
        ng.append("個人名らしき記述がある")

    print("=" * 40)
    if ng:
        for x in ng: print("  ⚠", x)
    else:
        print("  機械的なチェックは通過")
    print("=" * 40)
    print("""自分の目で確かめること
  □ 回答者の自由記述を、そのまま載せていないか
  □ 誰が書いたか分かってしまう内容が混じっていないか
  □ AIが例として作った文章を、自分の結果として載せていないか
  □ 「限界」に書いたことと、「結論」で言っていることが矛盾していないか""")

check()
```

> **3つめのチェックは、自分では気づきにくい。**
> AIに「参加者の声の例を作って」と頼むと、AIは**それらしい文章を作る**。
> それを自分の調査結果として載せれば、**ねつ造**になる。
