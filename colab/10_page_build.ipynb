## 1. 構成の型

発表も報告書もページも、骨は同じである。

```
問い  →  方法  →  結果  →  限界  →  結論
```

**「限界」を飛ばさないこと。**ここを書けるかどうかで、読む人の信頼が決まる。
「何が言えないか」を自分で言える人の「言えること」は、信じてもらえる。

### 1セクション＝1メッセージ

見出しをつけたら、その下に書くのは**1つのこと**だけである。
2つ言いたくなったら、セクションを分ける。

## 2. ページの中身を書く

下の `page` の中身を、自分の調査のものに書き換える。
**日本語を書くだけでよい。**

````python
page = {
    "title":   "ここにタイトル（何を調べたのかが分かる名前に）",
    "meta":    "情報活用　◯班（氏名／氏名／氏名／氏名）　2026年◯月◯日",

    # 何を確かめたかったのか。なぜ知りたかったのか
    "question": """
    ここに書く。
    """,

    # 誰に・いつ・どうやって聞いたか。回収数は必ず書く
    "method": """
    ここに書く。回収数は必ず書く。
    """,

    # 数字と図。読み取れることまで。意見はまだ書かない
    "result": """
    ここに書く。
    """,

    # 図のファイル名（第9回で savefig したもの）とキャプション
    "figures": [
        ("graph1.png", "図1　ここにキャプション（何のグラフか・n＝いくつか）"),
        # ("graph2.png", "図2　……"),
    ],

    # この調査で「言えないこと」
    "limits": """
    ここに書く。
    """,

    # 問いに対する答えを1〜2文で
    "conclusion": """
    ここに書く。
    """,

    # データの出どころ
    "source": "自分たちで実施したアンケート（2026年◯月／回収◯件）",
}
print("書き換えたら ▶ を押す")
````

````python
# ページを組み立てて表示する。▶ を押すだけ
import re
from IPython.display import HTML, display

def build(page, out="report.html", show_guide=False):
    """page の中身をテンプレートに流し込んで report.html を作る。"""
    tpl = open("template.html", encoding="utf-8").read()

    def paras(t):
        return "".join(f"<p>{line.strip()}</p>"
                       for line in t.strip().split("\n") if line.strip())

    figs = "\n".join(
        f'<figure><img src="{src}" alt="{cap}">'
        f'<figcaption>{cap}</figcaption></figure>'
        for src, cap in page["figures"])

    body = {
        "question":   paras(page["question"]),
        "method":     paras(page["method"]),
        "result":     paras(page["result"]),
        "limits":     paras(page["limits"]),
        "conclusion": paras(page["conclusion"]),
        "figures":    figs,
    }

    html = tpl
    html = html.replace("<h1>ここにタイトル</h1>", f"<h1>{page['title']}</h1>")
    html = html.replace("情報活用　◯班（氏名／氏名／氏名／氏名）　2026年◯月◯日", page["meta"])
    html = html.replace("自分たちで実施したアンケート（2026年◯月／回収◯件）", page["source"])

    for key, content in body.items():
        html = re.sub(rf'<div data-fill="{key}">.*?</div>',
                      f'<div data-fill="{key}">{content}</div>',
                      html, count=1, flags=re.S)

    if not show_guide:   # 書き方の案内文（灰色の小さい文字）は完成版から消す
        html = re.sub(r'<p class="lead">.*?</p>', "", html, flags=re.S)

    open(out, "w", encoding="utf-8").write(html)
    return out

build(page)
display(HTML(open("report.html", encoding="utf-8").read()))
````

---
## 3. できたHTMLを、いちど見てみる

下のセルを実行すると、**いま作られたHTMLの中身**が見える。

見てほしいのは**見出しの入れ子**である。

```html
<h1>タイトル</h1>
  <h2>問い</h2>
    <p>本文</p>
  <h2>方法</h2>
    <p>本文</p>
```

**この構造が、そのまま文章の構造である。**
Wordの「見出し1・見出し2」も、報告書の「章・節」も、同じことをしている。
だから **HTMLを覚える必要はない。構造を持っていれば、形式は移せる。**

````python
print(open("report.html", encoding="utf-8").read()[:1200])
````
